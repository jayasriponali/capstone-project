# Section 2 - Apply Missing Value Handling

```python
# find columns where missing percentage is less than 5
low_miss = []
for col_name, pct_val in null_pct.items():
    if pct_val <= 5 and pct_val != 0:
        print(f"column: {col_name}  missing: {pct_val:.4f}%")
        low_miss.append(col_name)
```

# Under 5% missing - drop those rows

embarked and embark_town both have 0.22% missing values.
Since the count is very small, we just drop the rows that have null in these columns.

```python
# drop rows where embarked or embark_town is null
print("Columns to drop rows for:", low_miss)
for col_name in low_miss:
    titanic_df.dropna(subset=[col_name], inplace=True)

# check that no nulls remain in those columns
print(titanic_df[titanic_df['embarked'].isna()])
print(titanic_df[titanic_df['embark_town'].isna()])
```

# 5% to 30% missing - impute

```python
# find columns where missing percentage is between 5 and 30
mid_miss = []
for col_name, pct_val in null_pct.items():
    if pct_val > 5 and pct_val <= 30:
        print(f"column: {col_name}  missing: {pct_val:.4f}%")
        mid_miss.append(col_name)

mid_miss
```

```python
# impute age with median since it is a numeric column
titanic_df['age'] = titanic_df['age'].fillna(titanic_df['age'].median())
```

```python
# verify that no nulls remain in age
titanic_df[titanic_df['age'].isna()]
```

# Reason for dropping the deck column

## 1) More than 77% of the values are missing. Imputing that much data would add a lot of noise.
## 2) If we drop all rows with null deck values, we would lose about 77% of the dataset.
## 3) Encoding missing as a category would give 77% of rows the same label which is not useful.

```python
titanic_df = titanic_df.drop(columns=['deck'])
```

```python
titanic_df.info()
```

# Justification

### embarked and embark_town had 0.22% missing values. We dropped those 2 rows since the amount is negligible.

### age had 19.87% missing values. We imputed with the median since dropping 20% of rows would be a big loss and median is a good estimate for numeric data with outliers.

### deck had 77.22% missing values. We dropped the column because imputing 77% of data would introduce too much noise and there is no good way to fill that many values.

# IQR Outlier Detection

# Outliers can be calculated for numeric columns only

```python
# formula for outliers
# Lower Bound = Q1 - 1.5 * IQR
# Upper Bound = Q3 + 1.5 * IQR

# Q1 = titanic_df['col'].quantile(0.25)
# Q3 = titanic_df['col'].quantile(0.75)
# IQR = Q3 - Q1
```

```python
# get all numeric columns
numeric_cols = titanic_df.select_dtypes(include=['number']).columns
print(f"Numeric columns : {numeric_cols}")
```

```python
outlier_dict = {}

for col in numeric_cols:
    Q1 = titanic_df[col].quantile(0.25)
    Q3 = titanic_df[col].quantile(0.75)
    IQR = Q3 - Q1
    print(f"column: {col}  IQR: {IQR}")

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"Lower Bound: {lower_bound}")
    print(f"Upper Bound: {upper_bound}")

    # count how many values fall outside the bounds
    outlier_dict[col] = ((titanic_df[col] < lower_bound) | (titanic_df[col] > upper_bound)).sum()

# print outlier counts for each column
for col, count in outlier_dict.items():
    print(f"{col}: {count} outliers")
```

#### Age outliers before and after imputation

```python
# The counts above are from the cleaned dataframe where the missing ages
# have already been filled with the median. Filling adds many values at
# the centre which changes the IQR and can shift the outlier count.
# So we also check the original age values from the saved CSV.

raw_age = pd.read_csv("titanic.csv")['age'].dropna()

Q1 = raw_age.quantile(0.25)
Q3 = raw_age.quantile(0.75)
IQR = Q3 - Q1

raw_lower = Q1 - 1.5 * IQR
raw_upper = Q3 + 1.5 * IQR

raw_outliers = ((raw_age < raw_lower) | (raw_age > raw_upper)).sum()

print(f"age before imputation: IQR range [{Q1}, {Q3}] -> {raw_outliers} outliers")
print(f"age after imputation: IQR range "
      f"[{titanic_df['age'].quantile(0.25)}, {titanic_df['age'].quantile(0.75)}]"
      f" -> {outlier_dict['age']} outliers")
```

# Section 4 - Bivariate Analysis

### Using boolean masking to compute survival rate by sex, pclass, and both together

# (a) Survival rate by sex

```python
male_mask = titanic_df['sex'] == 'male'
survived_mask = titanic_df['survived'] == 1

male_survived_mask = male_mask & survived_mask

total_male = titanic_df[male_mask].shape[0]
print(f"Total Male passengers: {total_male}")

total_male_survived = titanic_df[male_survived_mask].shape[0]
print(f"Total Male Survivors: {total_male_survived}")

male_survival_rate = (total_male_survived / total_male) * 100
print(f"Male survival rate: {male_survival_rate:.2f}")
```

```python
female_mask = titanic_df['sex'] == 'female'
survived_mask = titanic_df['survived'] == 1

female_survived_mask = female_mask & survived_mask

total_female = titanic_df[female_mask].shape[0]
print(f"Total Female passengers: {total_female}")

total_female_survived = titanic_df[female_survived_mask].shape[0]
print(f"Total Female Survivors: {total_female_survived}")

female_survival_rate = (total_female_survived / total_female) * 100
print(f"Female survival rate: {female_survival_rate:.2f}")
```

```python
print(f"Male survival rate: {male_survival_rate:.2f}")
print(f"Female survival rate: {female_survival_rate:.2f}")
```

# (b) Survival rate by pclass

#### For pclass we have multiple values, so get the unique values first
#### Then loop over each class to compute the survival rate

```python
for p_class in titanic_df['pclass'].unique():
    class_mask = titanic_df['pclass'] == p_class
    survived_mask = titanic_df['survived'] == 1
    survived_in_class = class_mask & survived_mask

    num_survived_in_class = titanic_df[survived_in_class].shape[0]
    total_in_class = titanic_df[class_mask].shape[0]

    class_rate = (num_survived_in_class / total_in_class) * 100
    print(f"Survival rate for class {p_class}: {class_rate:.2f}")
```

# (c) Survival rate by sex and pclass together

```python
for sex in titanic_df['sex'].unique():
    for p_class in titanic_df['pclass'].unique():
        sex_class_mask = ((titanic_df['pclass'] == p_class) & (titanic_df['sex'] == sex))
        survived_mask = ((titanic_df['survived'] == 1) & (titanic_df['pclass'] == p_class) & (titanic_df['sex'] == sex))

        total_group = titanic_df[sex_class_mask].shape[0]
        total_survived = titanic_df[survived_mask].shape[0]

        group_rate = (total_survived / total_group) * 100
        print(f"Sex: {sex.capitalize()}, Class: {p_class}, Total: {total_group}, Survived: {total_survived}, Rate: {group_rate:.2f}")
```

# Correlation Matrix - restricted to 6 numeric columns

### Using: survived, pclass, age, sibsp, parch, fare
### adult_male and alone are excluded because they are derived columns.
### adult_male is computed from sex and age, alone is computed from sibsp + parch.
### So they are not independent features.

```python
corr_cols = [
    'survived',
    'pclass',
    'age',
    'sibsp',
    'parch',
    'fare'
]

corr_matrix = titanic_df[corr_cols].corr()
```

# Draw the Heatmap

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    linewidths=0.7
)

plt.title('Correlation Matrix')
plt.show()
```

#### Find the Two Strongest Correlations

```python
abs_corr = corr_matrix.abs()

# use upper triangle to avoid duplicates
upper_tri = abs_corr.where(
    np.triu(np.ones(abs_corr.shape), k=1).astype(bool)
)

# sort and get the top 2
top_pairs = upper_tri.stack().sort_values(ascending=False)

print(top_pairs.head(2))
```

#### 1) The strongest correlation is between pclass and fare with a coefficient around -0.55.
#### This means that first class passengers paid much higher fares than third class.
#### As the class number goes up (1 to 3), the fare goes down, giving a strong negative correlation.

#### 2) The second strongest is between survived and pclass with a coefficient around -0.34.
#### First class passengers had better access to lifeboats and survived more.
#### Higher class number means lower survival rate, so the correlation is negative.
